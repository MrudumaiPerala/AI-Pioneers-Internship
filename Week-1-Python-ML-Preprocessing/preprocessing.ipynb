# Week 1: Python Data Preprocessing & Exploratory Data Analysis

## Objectives
- Missing value treatment (median for numerical, mode for categorical)
- Duplicate record removal
- Feature selection and schema alignment
- One-hot encoding of categorical variables (`pd.get_dummies`)
- Z-score feature normalization
- Correlation analysis and exploratory summary statistics

```python
import numpy as np
import pandas as pd
from pathlib import Path

# Load sample dataset
data_path = Path('data/sample_dataset.csv')
df = pd.read_csv(data_path)
print('Initial Dataset:')
display(df)
print('\nMissing values:\n', df.isnull().sum())
```

```python
# Missing Value Imputation
numeric_cols = ['Age', 'Salary', 'Experience']
for col in numeric_cols:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].median())

for col in ['Gender', 'City']:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

print('After Imputation Missing Values:\n', df.isnull().sum())
display(df)
```

```python
# Remove duplicates & select features
df = df.drop_duplicates()
selected_features = ['Age', 'Salary', 'Experience', 'Gender', 'Department', 'Target']
df = df[selected_features]

# One-Hot Encoding
encoded_df = pd.get_dummies(df, columns=['Gender', 'Department'], drop_first=True)
display(encoded_df)
```

```python
# Normalization & Export
X = encoded_df.drop(columns=['Target'])
numeric_features = X.select_dtypes(include=[np.number]).columns.tolist()
X_scaled = X.copy()
for col in numeric_features:
    mean = X_scaled[col].mean()
    std = X_scaled[col].std(ddof=0)
    X_scaled[col] = (X_scaled[col] - mean) / std if std != 0 else 0

encoded_df.to_csv('data/cleaned_dataset.csv', index=False)
print('Cleaned dataset saved to data/cleaned_dataset.csv')
display(encoded_df.describe().T)
```
